### Task 2: Stakeholders and Hypotheses — Conclusion

The FreshMart investigation involves several stakeholders, including executive management, finance, marketing, store operations, regional managers, supply chain management and customers. Each stakeholder has a different interest in improving business performance, but all are affected by changes in customer purchasing behaviour and revenue generation.

Six competing hypotheses were developed to investigate the potential reasons for declining basket size. These include pricing and affordability pressures, increased competition, product availability issues, changing shopping behaviour, changes in customer mix and differences in store formats and operational performance.

The Transactions, Stores and Stockouts datasets provide information that can be used to investigate these hypotheses through customer segmentation, store comparisons, time-based analysis and operational analysis.

Each hypothesis will be tested against measurable evidence to determine whether the data supports or rejects the proposed explanation. The investigation will also consider alternative explanations to avoid confusing changes in customer composition with genuine changes in individual customer purchasing behaviour.

The findings will provide a foundation for identifying the most likely causes of FreshMart's declining basket size and developing evidence-based business recommendations.

```python
# Load the FreshMart Transactions table
# (the data dictionary only describes fields; the dates live in the transactions data)
Transactions = spark.table("workspace.pandas.fresh_mart_dataset_1").toPandas()

# Convert transaction dates into datetime
Transactions['transaction_date'] = pd.to_datetime(
    Transactions['transaction_date'],
    errors='coerce'
)

# Display earliest and latest transaction dates
print("First transaction:",
      Transactions['transaction_date'].min())

print("Last transaction:",
      Transactions['transaction_date'].max())
```
